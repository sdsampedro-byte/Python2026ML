```python
# Creación aleatoria de datos
import numpy as np
import random
random.seed(42)
```

```python
# Creo una variable de tipo lista vacía
nivel_estudios = []
# Creo o ingreso a mi lista la variable Secuandaria y creo 110 veces secundaria, Bachillerato 172, etc.
nivel_estudios.extend(['Secundaria']*110)
nivel_estudios.extend(['Bachillerato']*172)
nivel_estudios.extend(['Grado']*145)
nivel_estudios.extend(['Master']*52)
nivel_estudios.extend(['Doctorado']*21)
# Mezcla y repite los datos de forma randomica
nivel_estudios = np.random.permutation(nivel_estudios)
```

```python
print(nivel_estudios)
```

```python
# DIAGRAMA DE BARRAS
import matplotlib.pyplot as plt
import numpy as np
estudios, frecuencia = np.unique(nivel_estudios, return_counts=True)
print(estudios, frecuencia)
```

```python
plt.ylabel('Frecuencia absoluta')
plt.xlabel('Nivel de estudios')
plt.title('Diagrama de barras')
plt.bar(estudios, frecuencia) # https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.bar.html
plt.show()
```

```python
# DIAGRAMA DE SECTORES
colores = {'red', 'blue', 'orange', 'green', 'cyan'}
plt.pie(frecuencia, labels=estudios, colors=colores, autopct='%1.2f%%')
plt.title('Diagrama de sectores')
plt.show()
```

```python
# Tarea implementar con pandas esta información
import numpy as np
import pandas as pd
import random

random.seed(42)

# Recrear los datos
nivel_estudios = []
nivel_estudios.extend(['Secundaria'] * 110)
nivel_estudios.extend(['Bachillerato'] * 172)
nivel_estudios.extend(['Grado'] * 145)
nivel_estudios.extend(['Master'] * 52)
nivel_estudios.extend(['Doctorado'] * 21)
nivel_estudios = np.random.permutation(nivel_estudios)

# Crear DataFrame
df = pd.DataFrame({'nivel_estudios': nivel_estudios})

# Mostrar primeras filas
print(df.head(10))

# Tabla de frecuencias
frecuencia_df = df['nivel_estudios'].value_counts().reset_index()
frecuencia_df.columns = ['Nivel de Estudios', 'Frecuencia']
print(frecuencia_df)
```

```python
# DIAGRAMA DE BARRAS con pandas
ax = frecuencia_df.plot(
    kind='bar',
    x='Nivel de Estudios',
    y='Frecuencia',
    color='steelblue',
    legend=False,
    title='Diagrama de Barras'
)
ax.set_xlabel('Nivel de Estudios')
ax.set_ylabel('Frecuencia Absoluta')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
